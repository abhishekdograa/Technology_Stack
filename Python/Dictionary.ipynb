```python
my_test={}
print(type(my_test))

my_test1=dict()
print(type(my_test1))

attr_dict={"name":"Abhi","age":19}
print(attr_dict)
print(type(attr_dict))
```

```python
for key in attr_dict:
    print(key)

for value in attr_dict.values():
    print(value)

for a,b in attr_dict.items():
    print(a,b)
```

```python
# Create a dict with default values
deft=dict.fromkeys(["name","age"],"NA")
print(deft)
```

```python
# Get the value associated with a key using the dict.get() method with a default value.
print(attr_dict.get("name"),"shek")
```
